```python
df_numeric = df_tratado.select_dtypes(include="number")

pearson_correlation = df_numeric.corr(method="pearson")

plt.figure(figsize=(18, 11))

sns.heatmap(
    pearson_correlation,
    annot=True,
    cmap="coolwarm",
    fmt=".2f",
    vmin=-1,
    vmax=1
)

plt.show()
```

```python
df_numeric = df_tratado.select_dtypes(include="number")

spearman_correlation = df_numeric.corr(method="spearman")

plt.figure(figsize=(18, 11))

sns.heatmap(
    spearman_correlation,
    annot=True,
    cmap="coolwarm",
    fmt=".2f",
    vmin=-1,
    vmax=1
)

plt.show()
```

```python
mask = np.triu(np.ones(pearson_correlation.shape, dtype=bool), k=1)

biggest_corr = (
    pearson_correlation.where(mask)
    .stack()
    .sort_values(key=abs, ascending=False)
    .head(10)
)

print(biggest_corr)
```

```python
mask = np.triu(np.ones(spearman_correlation.shape, dtype=bool), k=1)

biggest_corr = (
    spearman_correlation.where(mask)
    .stack()
    .sort_values(key=abs, ascending=False)
    .head(10)
)

print(biggest_corr)
```
