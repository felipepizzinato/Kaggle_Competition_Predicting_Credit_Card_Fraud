```python
df_numeric = df_tratado.select_dtypes(include="number")

correlacao = df_numeric.corr(method="pearson")

plt.figure(figsize=(10, 6))

sns.heatmap(
    correlacao,
    annot=True,
    cmap="coolwarm",
    fmt=".2f",
    vmin=-1,
    vmax=1
)

plt.title("Matriz de Correlação")
plt.show()
```

```python
df_numeric = df_tratado.select_dtypes(include="number")

correlacao = df_numeric.corr(method="spearman")

plt.figure(figsize=(10, 6))

sns.heatmap(
    correlacao,
    annot=True,
    cmap="coolwarm",
    fmt=".2f",
    vmin=-1,
    vmax=1
)

plt.title("Matriz de Correlação")
plt.show()
```
